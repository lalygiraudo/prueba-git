```python
# ============================================
# EDA - DATASET IRIS
# ============================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Cargar dataset público
df = sns.load_dataset("iris")

# Mostrar las primeras filas
df.head()
```

```python
# Verificación de valores faltantes
df.isnull().sum()
```

```python
# En porcentaje
porcentaje_nulos = df.isnull().mean() * 100

porcentaje_nulos
```

```python
# Cuantas flores de cada especie hay
df["species"].value_counts()
```

```python
# Visualizaciones
sns.countplot(
    data=df,
    x="species"
)

plt.title("Cantidad de observaciones por especie")
plt.xlabel("Especie")
plt.ylabel("Cantidad")

plt.show()
```

```python
# Histograma
df.hist(
    figsize=(12, 8),
    bins=15
)

plt.tight_layout()
plt.show()
```

¿Todas las variables tienen una distribución similar?

¿Cuál parece tener mayor dispersión?

¿Observamos valores extraños?

```python
# Boxplots
plt.figure(figsize=(10, 6))

sns.boxplot(data=df)

plt.title("Distribución de las variables")
plt.xticks(rotation=45)

plt.show()
```

```python
# Correlación
correlacion = df.drop(columns="species").corr()

correlacion
```

```python
# Matriz de correlación
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlacion,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Matriz de correlación")

plt.show()
```

¿Qué variables presentan una relación más fuerte?

```python
# Relación entre dos variables
sns.scatterplot(
    data=df,
    x="petal_length",
    y="petal_width",
    hue="species"
)

plt.title("Largo y ancho del pétalo")
plt.xlabel("Largo del pétalo")
plt.ylabel("Ancho del pétalo")

plt.show()
```
