```python
# ============================================================
# CLASIFICACIÓN BINARIA CON REGRESIÓN LOGÍSTICA
# DATASET: BREAST CANCER WISCONSIN
# ============================================================

# 1. IMPORTAR LIBRERÍAS

import pandas as pd

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
```

```python
# ============================================================
# 2. CARGAR EL DATASET
# ============================================================

datos = load_breast_cancer()

X = pd.DataFrame(
    datos.data,
    columns=datos.feature_names
)

y = pd.Series(
    datos.target,
    name="diagnostico"
)

print("Cantidad de filas:", X.shape[0])
print("Cantidad de variables:", X.shape[1])
```

```python
# ============================================================
# 3. CONOCER LOS DATOS
# ============================================================

print("Primeras filas:")

display(X.head())
```

```python
# ============================================================
# 4. CONOCER LA VARIABLE OBJETIVO
# ============================================================

print("Clases disponibles:")

print(datos.target_names)

print("\nCantidad de casos por clase:")

print(y.value_counts())
```

0 = malignant
1 = benign

```python
# ============================================================
# 5. SEPARAR DATOS DE ENTRENAMIENTO Y PRUEBA
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Datos de entrenamiento:", X_train.shape)
print("Datos de prueba:", X_test.shape)
```

```python
# ============================================================
# 6. ESTANDARIZAR LAS VARIABLES
# ============================================================

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)
```

```python
# ============================================================
# 7. CREAR EL MODELO
# ============================================================

modelo = LogisticRegression(
    max_iter=1000,
    random_state=42
)

# Entrenar
modelo.fit(X_train_scaled, y_train)

print("Modelo entrenado correctamente.")
```

```python
# ============================================================
# 8. REALIZAR PREDICCIONES
# ============================================================

y_pred = modelo.predict(X_test_scaled)

print("Predicciones:")
print(y_pred)

print("\nValores reales:")
print(y_test.values)
```

```python
# ============================================================
# 9. ACCURACY
# ============================================================

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", round(accuracy, 3))
```

```python
# ============================================================
# 10. MATRIZ DE CONFUSIÓN
# ============================================================

matriz = confusion_matrix(y_test, y_pred)

print(matriz)
```

```python
print("             Predicción")
print("             Maligno  Benigno")
print()
print("Real Maligno", matriz[0])
print("Real Benigno", matriz[1])
```

```python
# ============================================================
# 11. REPORTE DE CLASIFICACIÓN
# ============================================================

print(
    classification_report(
        y_test,
        y_pred,
        target_names=datos.target_names
    )
)
```

```python
# ============================================================
# 12. PROBABILIDADES
# ============================================================

probabilidades = modelo.predict_proba(X_test_scaled)

print("Primeros 5 casos:")
print()

for i in range(5):
    print(
        f"Caso {i+1}: "
        f"Maligno = {probabilidades[i][0]:.2%} | "
        f"Benigno = {probabilidades[i][1]:.2%}"
    )
```

¿ Qué pasó con el desbalanceo?

El desbalanceo existe, pero no es suficientemente extremo como para impedir que la Regresión Logística funcione bien, y además estamos evaluando con varias métricas, no solamente con accuracy.

En este dataset hay aproximadamente:

357 casos benignos
212 casos malignos

Equivale aproximadamente a 63% vs. 37%. Hay desbalance, pero no es un caso extremo como 99% vs. 1%.

Lo importante: ¿qué pasaría si el modelo simplemente eligiera la clase mayoritaria?

En clasificación, accuracy sola puede engañarnos.


En este problema, un falso negativo (FN) significa:

El modelo predijo "benigno" cuando realmente era "maligno".

Ese error puede ser mucho más relevante que un falso positivo.

Por eso conviene mirar:

* Precision
* Recall
* F1-score
* Matriz de confusión

¿Y por qué usamos `stratify=y`?

Esta línea:

train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

es importante.

`stratify=y` hace que entrenamiento y prueba mantengan aproximadamente la misma proporción de clases que el dataset original.

Sin `stratify`, podríamos tener, por azar, una distribución diferente entre train y test.

Entonces, ¿hay que corregir el desbalanceo?

No necesariamente.

Detectar desbalanceo no significa automáticamente aplicar SMOTE, undersampling o oversampling.

Primero hay que evaluar qué tan desbalanceadas están las clases y qué impacto tiene sobre el modelo.
